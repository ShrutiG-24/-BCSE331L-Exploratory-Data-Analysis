## Data Handling and Cleaning

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

```python
df_clean = df.copy()
```

```python
# Mask for non-empty values
mask = df_clean['INCOME'].notna()

# remove brackets,parentheses
df_clean.loc[mask, 'INCOME'] = (
    df_clean.loc[mask, 'INCOME']
    .astype(str)
    .str.replace(r'[\[\]\(\)]', '', regex=True)
    .str.strip()
)
# remove $ signs, commas,dots
df_clean.loc[mask, 'INCOME'] = (
    df_clean.loc[mask, 'INCOME']
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.replace('.', '', regex=False)
)

# numeric ranges
income_split = df_clean.loc[mask, 'INCOME'].str.extract(r'(\d+)-?(\d+)?')
df_clean['INCOME_MIN'] = pd.to_numeric(income_split[0], errors='coerce')
df_clean['INCOME_MAX'] = pd.to_numeric(income_split[1], errors='coerce')

# starts with "-" → min=0
starts_with_dash = df_clean.loc[mask, 'INCOME'].str.startswith('-')
df_clean.loc[starts_with_dash, 'INCOME_MIN'] = 0
df_clean.loc[starts_with_dash, 'INCOME_MAX'] = pd.to_numeric(
    df_clean.loc[starts_with_dash, 'INCOME'].str.replace('-', '', regex=False),
    errors='coerce'
)

# starts with "-" → max = column max
ends_with_dash = df_clean.loc[mask, 'INCOME'].str.endswith('-')
col_max = df_clean['INCOME_MAX'].max(skipna=True)
df_clean.loc[ends_with_dash, 'INCOME_MIN'] = pd.to_numeric(
    df_clean.loc[ends_with_dash, 'INCOME'].str.replace('-', '', regex=False),
    errors='coerce'
)
df_clean.loc[ends_with_dash, 'INCOME_MAX'] = col_max

# max -> fill with min
df_clean['INCOME_MAX'] = df_clean['INCOME_MAX'].fillna(df_clean['INCOME_MIN'])

# Average for clean value
df_clean['INCOME_CLEAN'] = df_clean[['INCOME_MIN', 'INCOME_MAX']].mean(axis=1)

df_clean['Income Range'] = df_clean.apply(
    lambda row: f"{int(row['INCOME_MIN'])}-{int(row['INCOME_MAX'])}"
    if pd.notna(row['INCOME_MIN']) and pd.notna(row['INCOME_MAX'])
    else np.nan,
    axis=1
)
print(df_clean[['INCOME_MIN', 'INCOME_MAX', 'Income Range']]
      .sample(10, random_state=42)
      .to_string(index=False))

```

## Multivariate Analysis

```python
df1 = df_clean.copy()
df1 = df1.drop(columns=['INCOME','INCOME_MIN','INCOME_MAX','INCOME_CLEAN','Income Range'])


sex_map = {'F': 0, 'M': 1}
marital_map = {'Single': 0, 'Married': 1, 'Divorced': 2, 'Together': 3, 'Widowed': 4}
age_map = {'14-17': 0, '18-24': 1, '25-34': 2, '35-44': 3, '45-54': 4, '55-64': 5, '65+': 6}
education_map = {'Grade 8 or less': 0, 'Grades 9 to 11': 1, 'Graduated High Scool': 2,
                 '1 to 3 years of college': 3, 'College graduate': 4, 'Grad Study': 5}
occupation_map = {'Homemaker': 0, 'Professional/Managerial': 1, 'Student, HS or College': 2,
                  'Retired': 3, 'Unemployed': 4, 'Factory Worker/Laborer/Driver': 5,
                  'Sales Worker': 6, 'Clerical/Service Worker': 7, 'Military': 8}
area_map = {'-1 year': 0, '1-3 years': 1, '4-6 years': 2, '7-10 years': 3, '10+ years': 4}
dual_map = {'No': 0, 'Yes': 1, 'Not Married': 2}
household_size_map = {'One': 1, 'Two': 2, 'Three': 3, 'Four': 4, 'Five': 5,
                      'Six': 6, 'Seven': 7, 'Eight': 8, 'Nine or more': 9}
under18_map = {'One': 1, 'Two': 2, 'Three': 3, 'Four': 4, 'Five': 5,
               'Six': 6, 'Seven': 7, 'Eight': 8, 'Nine or more': 9}
householder_map = {'Own': 0, 'Rent': 1, 'Family': 2}
home_type_map = {'House': 0, 'Apartment': 1, 'Other': 2, 'Condominium': 3, 'Mobile Home': 4}
ethnic_map = {'White': 0, 'Hispanic': 1, 'Asian': 2, 'Black': 3, 'Other': 4,
              'East Indian': 5, 'American Indian': 6, 'Pacific Islander': 7}
language_map = {'English': 0, 'Spanish': 1, 'Other': 2}

# Apply mappings
df1['SEX'] = df1['SEX'].map(sex_map)
df1['MARITAL.STATUS'] = df1['MARITAL.STATUS'].map(marital_map)
df1['AGE'] = df1['AGE'].map(age_map)
df1['EDUCATION'] = df1['EDUCATION'].map(education_map)
df1['OCCUPATION'] = df1['OCCUPATION'].map(occupation_map)
df1['AREA'] = df1['AREA'].map(area_map)
df1['DUAL.INCOMES'] = df1['DUAL.INCOMES'].map(dual_map)
df1['HOUSEHOLD.SIZE'] = df1['HOUSEHOLD.SIZE'].map(household_size_map)
df1['UNDER18'] = df1['UNDER18'].map(under18_map)  # NaNs stay NaN
df1['HOUSEHOLDER'] = df1['HOUSEHOLDER'].map(householder_map)
df1['HOME.TYPE'] = df1['HOME.TYPE'].map(home_type_map)
df1['ETHNIC.CLASS'] = df1['ETHNIC.CLASS'].map(ethnic_map)
df1['LANGUAGE'] = df1['LANGUAGE'].map(language_map)

corr_matrix = df1.corr()

plt.figure(figsize=(15, 10))
sns.heatmap(corr_matrix, cmap='coolwarm', annot=True)
plt.title("Correlation Matrix (Integer Encoded) from df1")
plt.show()
```

### **1. Strongest correlations**
* **AGE & HOUSEHOLDER (-0.6):** Older individuals are less likely to rent or be in a "family" householder category. This is a strong negative correlation.
* **DUAL.INCOMES & HOUSEHOLDER (0.44):** Households with dual incomes are moderately positively correlated with certain householder types (likely "Own").
* **AGE & MARITAL.STATUS (0.42):** Older individuals are more likely to be married, which makes sense demographically.
* **AGE & EDUCATION (0.33):** Slight positive correlation — older age groups have higher education (or encoding order may reflect education level vs age group).

### **2. Low or negligible correlations**
* Most other variable pairs have correlations close to 0 (between -0.2 and 0.2), indicating weak or no linear relationship:
  * SEX is almost independent of other features.
  * ETHNIC.CLASS and LANGUAGE show very weak correlations with other columns.
  * AREA, OCCUPATION, HOME.TYPE have low correlations overall.

### **CONCLUSION**
* **Demographics drive correlation:** Age, marital status, education, and household characteristics show the strongest relationships.
* **Most variables are largely independent:** This is expected with categorical demographic features encoded as integers.

### Row & Column Profiles + Chi-square

```python
import pandas as pd
from scipy.stats import chi2_contingency

# Create contingency table
ct = pd.crosstab(df_clean["SEX"], df_clean["MARITAL.STATUS"])

print("Contingency Table:\n", ct)

# Row profiles: proportions across columns for each row
row_profiles = ct.div(ct.sum(axis=1), axis=0)
print("\nRow profiles:\n", row_profiles)

# Column profiles: proportions across rows for each column
col_profiles = ct.div(ct.sum(axis=0), axis=1)
print("\nColumn profiles:\n", col_profiles)

# Relative frequencies: proportion of total
rel_freq = ct / ct.values.sum()
print("\nRelative frequencies:\n", rel_freq)

# Chi-square test of independence
chi2, p, dof, expected = chi2_contingency(ct)
print(f"\nChi-square value: {chi2:.2f}")
print(f"p-value: {p:.4f}")
print(f"Degrees of freedom: {dof}")
print("Expected frequencies:\n", expected)
```

```python
ct2 = pd.crosstab(df_clean["SEX"], df_clean["EDUCATION"])
print("\nContingency Table: SEX vs EDUCATION\n", ct2)

row_profiles2 = ct2.div(ct2.sum(axis=1), axis=0)
print("\nRow profiles:\n", row_profiles2)

col_profiles2 = ct2.div(ct2.sum(axis=0), axis=1)
print("\nColumn profiles:\n", col_profiles2)

rel_freq2 = ct2 / ct2.values.sum()
print("\nRelative frequencies:\n", rel_freq2)

chi2_2, p2, dof2, expected2 = chi2_contingency(ct2)
print(f"\nChi-square value: {chi2_2:.2f}, p-value: {p2:.4f}")
```

```python
ct3 = pd.crosstab(df_clean["MARITAL.STATUS"], df_clean["HOME.TYPE"])
print("\nContingency Table: MARITAL.STATUS vs HOME.TYPE\n", ct3)

row_profiles3 = ct3.div(ct3.sum(axis=1), axis=0)
print("\nRow profiles:\n", row_profiles3)

col_profiles3 = ct3.div(ct3.sum(axis=0), axis=1)
print("\nColumn profiles:\n", col_profiles3)

rel_freq3 = ct3 / ct3.values.sum()
print("\nRelative frequencies:\n", rel_freq3)

chi2_3, p3, dof3, expected3 = chi2_contingency(ct3)
print(f"\nChi-square value: {chi2_3:.2f}, p-value: {p3:.4f}")
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
df_income = df_clean[["INCOME_CLEAN"]]
scaler = StandardScaler()
income_scaled = scaler.fit_transform(df_income)

optimal_k = 3
kmeans = KMeans(n_clusters=optimal_k, random_state=42)
df_clean["IncomeCluster"] = kmeans.fit_predict(income_scaled)

plt.figure(figsize=(8, 6))
plt.scatter(df_clean.index, df_clean["INCOME_CLEAN"],c=df_clean["IncomeCluster"], cmap='viridis')
plt.xlabel("Index")
plt.ylabel("INCOME_CLEAN")
plt.title("K-Means Clusters on Income")
plt.show()

centroids = scaler.inverse_transform(kmeans.cluster_centers_)
print("\nCluster centroids (Income):\n", centroids)

```

```python
def age_midpoint(age_range):
    try:
        parts = age_range.split('-')
        return (float(parts[0]) + float(parts[1])) / 2
    except:
        return np.nan

df_clean["AGE_NUM"] = df_clean["AGE"].apply(age_midpoint)
df_clean["AGE_NUM"].fillna(df_clean["AGE_NUM"].mean(), inplace=True)
df_age = df_clean[["AGE_NUM"]]
scaler = StandardScaler()
age_scaled = scaler.fit_transform(df_age)
optimal_k = 4
kmeans = KMeans(n_clusters=optimal_k, random_state=42)
df_clean["AgeCluster"] = kmeans.fit_predict(age_scaled)
plt.figure(figsize=(8, 6))
plt.scatter(df_clean.index, df_clean["AGE_NUM"],c=df_clean["AgeCluster"], cmap='viridis')
plt.xlabel("Index")
plt.ylabel("AGE_NUM")
plt.title("K-Means Clusters on Age")
plt.show()

centroids = scaler.inverse_transform(kmeans.cluster_centers_)
print("\nCluster centroids (Age):\n", centroids)


```
